```python
import pandas as pd
df = pd.read_csv("emails.csv")
print(df.head())
```

```python
df['Email Type'].value_counts()
```

```python
df = df.rename(columns = {"Email Type":"label","Email Text":"text"})
```

```python
df.info()
```

```python
df.shape
```

```python
df.dropna(subset = ['text'],inplace = True)
```

```python
df.shape
```

```python
df.drop(columns = ['Unnamed: 0'], inplace = True)
```

```python
df.info()
```

```python
df['label']  = df['label'].map({'Safe Email':0,'Phishing Email':1})
```

```python
df['label'].value_counts()
```

```python
import re

def clean_text(text):
    if pd.isnull(text):
        return ""
    text = text.lower()
    text = re.sub(r'[^\w\s:/.-]', '', text)  # retain URL & number structure
    text = re.sub(r'\s+', ' ', text).strip()
    return text
df['text'] = df['text'].apply(clean_text)
```

```python
df.to_csv("emails_cleaned.csv",index = False)
```
